# Difference Equations

Economists often study how economic variables move over time, such as national income, interest rates, money supply, production of oil, price of rice, and so on. Time can be treated as either discrete or continuous. A useful way to model the dynamics of variables of interest at discrete time intervals (e.g., daily, weekly, monthly, or annually) is through difference equations or recurrence relations.

## First-Order Difference Equations

The first-order linear difference equation can be written generally as:

$$x_{t} = ax_{t-1} + c$$

::: {.callout-note}
**Key Idea**

A difference equation describes how a variable evolves **step by step over time**. Instead of continuous change (as in differential equations), time moves in discrete periods: $t=0,1,2,\dots$.
:::

### Example 1: Convergent and Monotonic

Consider the following equation with a starting value $x_{0} = 400$:

:::{math}
:enumerated: false
x_{t} = \frac{1}{2}x_{t-1} + 300
:::

The first few terms are calculated as:
* $x_{1} = \frac{1}{2}(400) + 300 = 500$
* $x_{2} = \frac{1}{2}(500) + 300 = 550$
* $x_{3} = \frac{1}{2}(550) + 300 = 575$
* $x_{4} = \frac{1}{2}(575) + 300 = 587.5$

The series $x_{t}$ increases at a decreasing rate. To find the steady state ($x_{\infty}$), we set $x_{t} = x_{t-1} = x_{\infty}$:

:::{math}
:enumerated: false
x_{\infty} = \frac{c}{1-a} = \frac{300}{1 - 0.5} = 600
:::

This path is **monotonic** because the sign of $a > 0$ (positive), and it is **stable** (convergent) because $|a| < 1$.

```python
import matplotlib.pyplot as plt

def simulate_multiple_paths(a, c, x_starts, t=10):
    plt.figure(figsize=(10, 6))
    
    # Calculate Steady State for reference
    # x* = c / (1 - a)
    if a != 1:
        steady_state = c / (1 - a)
        plt.axhline(steady_state, color='black', linestyle='--', alpha=0.7, label=f'Steady State ({steady_state:.1f})')
    
    for x0 in x_starts:
        path = [x0]
        for _ in range(t):
            path.append(a * path[-1] + c)
        plt.plot(path, 'o-', label=f'$x_0 = {x0}$')

    plt.title(f"Difference Equation Paths: $x_t = {a}x_{{t-1}} + {c}$")
    plt.xlabel("Time (t)")
    plt.ylabel("$x_t$")
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# --- SCENARIO 1: Stable & Monotonic (Page 1-2) ---
# a = 0.5
simulate_multiple_paths(a=0.5, c=300, x_starts=[100, 400, 600, 900])

# --- SCENARIO 2: Unstable & Monotonic (Page 3) ---
# a = -0.5 
# simulate_multiple_paths(a=-0.5, c=300, x_starts=[-400, 100, 200, 400], t=5)

# --- SCENARIO 3: Unstable & Monotonic (Page 4) ---
# a = 2.0
# simulate_multiple_paths(a=2.0, c=-300, x_starts=[-200, 0, 300, 500], t=6)

# --- SCENARIO 4: Unstable & Oscillatory (Page 4) ---
# a = -2.0 (less than -1)
# simulate_multiple_paths(a=-2.0, c=900, x_starts=[250, 300, 350], t=6)
```

### Example 2: Convergent and Oscillatory

Consider $x_{t} = -\frac{1}{2}x_{t-1} + 300$ with $x_{0} = 400$:

* $x_{1} = -\frac{1}{2}(400) + 300 = 100$
* $x_{2} = -\frac{1}{2}(100) + 300 = 250$
* $x_{3} = -\frac{1}{2}(250) + 300 = 175$

This series is oscillatory because $a < 0$ (negative), but it still converges to a steady state of $x_{\infty} = \frac{300}{1 - (-0.5)} = 200$ because $|a| < 1$.

::: {.callout-tip}
**Economic Interpretation**

Stable difference equations often describe gradual adjustment in economics — for example, capital accumulation, price adjustment, or expectations updating.
:::

```python
# --- SCENARIO 2: Unstable & Monotonic (Page 3) ---
a = -0.5 
simulate_multiple_paths(a=-0.5, c=300, x_starts=[-400, 100, 200, 400], t=5)
```

### Example 3: Divergent and Monotonic

Consider $x_{t} = 2x_{t-1} - 300$ with $x_{0} = 400$:

* $x_{1} = 2(400) - 300 = 500$
* $x_{2} = 2(500) - 300 = 700$
* $x_{3} = 2(700) - 300 = 1100$
* $x_{4} = 2(1100) - 300 = 1900$

and so on.

The series diverges monotonically because $a > 0$ (positive). Note that if the series starts exactly at its steady state ($x_{\infty} = 300$), it remains there.

```python
# --- SCENARIO 3: Unstable & Monotonic (Page 4) ---
a = 2.0
simulate_multiple_paths(a=2.0, c=-300, x_starts=[-200, 0, 300, 500], t=6)
```

### Example 4: Divergent and Oscillatory

Consider $x_{t} = -2x_{t-1} + 900$ with $x_{0} = 400$:

* $x_{1} = -2(400) + 900 = 100$
* $x_{2} = -2(100) + 900 = 700$
* $x_{3} = -2(700) + 900 = -500$

This path is oscillatory and divergent because $a < 0$ (negative).

::: {.callout-note}
**Knife-Edge Equilibrium**

When $|a|>1$, the steady state is unstable. The system will only remain at that point if it starts there exactly — any small deviation grows over time.
:::

```python
# --- SCENARIO 4: Unstable & Oscillatory (Page 4) ---
a = -2.0
simulate_multiple_paths(a=-2.0, c=900, x_starts=[-100, 250, 0, 350], t=6)
```

## Summary of Stability

For a first-order linear difference equation:

:::{math}
:enumerated: false
x_{t} = ax_{t-1} + c
:::

we can determine the behavior of the series by calculating its steady state (equilibrium value), where $x_t = x_{t-1} = x_{\infty}$:

:::{math}
:enumerated: false
x_{\infty} = \frac{c}{1 - a} \quad (\text{for } a \neq 1)
:::

A first-order difference equation is stable if it converges to the steady state regardless of the starting value $x_{0}$. 

This occurs if:

* $|a| < 1$ or $-1 < a < 1$.
* And if $a > 0$, the path is monotonic.
* If $a < 0$, the path is oscillatory.

Conversely, if $|a| > 1$, the path is unstable and diverges - monotically if $a$ is positive and oscillatory if negative.

| Condition for $a$ | Path Characteristic | Convergence/Stability |
|------------------------|--------------------|-----------------------|
| $ 0 < a < 1 $        | Monotonic          | Stable (Converges to $ x^* $) |
| $ -1 < a < 0 $       | Oscillatory        | Stable (Converges to $ x^* $) |
| $ a > 1 $            | Monotonic          | Unstable (Diverges to $ \pm \infty $) |
| $ a < -1 $           | Oscillatory        | Unstable (Diverges/Explodes) |

### Special Cases

What happens when $a = 1$ and $a = -1$?

* $a = 1$

The equation becomes $x_t = x_{t-1} + c$. This results in a straight line $x_t = x_0 + tc$, which diverges monotonically. 

* $a = -1$

The equation becomes $x_t = -x_{t-1} + c$. Unless $x_0$ is exactly at the steady state ($c/2$), the values will oscillate between two constant points indefinitely. 

| Condition for $a$ | Path Characteristic | Convergence/Stability |
|------------------------|--------------------|-----------------------|
| $ a = 1 $            | Linear Trend       | Unstable ($ x_t = x_0 + tc $) |
| $ a = -1 $           | Fixed Oscillation  | Unstable (Cycles around $ x^* $) |

### Solution to First-Order Difference Equation by Repeated Iteration

The following derivation demonstrates how to find the general solution for a first-order linear difference equation by repeatedly substituting the equation into itself.

Starting with the general first-order linear difference equation:

:::{math}
:label: eq-difference
x_{t} = a x_{t-1} + c
:::

Since this relationship holds for any time period, we know that for period $t-1$:

:::{math}
:label: eq-difference-1
x_{t-1} = ax_{t-2} + c
:::

**Step 1: First Substitution**

Substitute Equation {eq}`eq-difference-1` into Equation {eq}`eq-difference`:

:::{math}
:enumerated: false
x_{t} = a(ax_{t-2} + c) + c
:::

:::{math}
:enumerated: false
x_{t} = a^{2}x_{t-2} + ac + c
:::

**Step 2: Second Substitution**

We also know that $x_{t-2} = ax_{t-3} + c$. Hence sbstituting this in:

:::{math}
:enumerated: false
x_{t} = a^{2}(ax_{t-3} + c) + ac + c
:::

:::{math}
:enumerated: false
x_{t} = a^{3}x_{t-3} + a^{2}c + ac + c
:::

**Step 3: Generalizing the Pattern**


Continuing this repeated substitution $t$ times, we arrive at the general solution by iteration:

:::{math}
:enumerated: false
x_{t} = a^{t}x_{0} + a^{t-1}c + a^{t-2}c + \dots + a^{2}c + ac + c
:::

This can be written more compactly using summation notation:

:::{math}
:enumerated: false
x_{t} = a^{t}x_{0} + \sum_{i=0}^{t-1} a^{i}c
:::

**Step 4: Using the Geometric Series Formula**

The term $\sum_{i=0}^{t-1} a^{i}$ is a geometric series. It can be shown that for $a \neq 1$:

:::{math}
:enumerated: false
\sum_{i=0}^{t-1} a^{i} = \frac{1-a^{t}}{1-a}
:::

Since the constant $c$ is not affected by the summation index $i$, we have $\sum_{i=0}^{t-1} a^{i}c = c\sum_{i=0}^{t-1} a^{i}$. Substituting the series formula into our solution gives:

:::{math}
:label: defn-solution-1
x_{t} = a^{t}x_{0} + \frac{1-a^{t}}{1-a}c
:::

::: {.callout-note}
**Reading the Solution**

The solution has two parts:

- $a^t x_0$ → the effect of the **initial condition**
- the remaining terms → the effect of the **forcing term**

So the time path reflects both where the system starts and how it is pushed over time.
:::

---

### Stability and the Steady State

This solution allows us to see exactly what happens as time ($t$) goes to infinity. 

* **If $|a| < 1$**: As $t \to \infty$, the term $a^{t}$ goes to $0$. 
    The equation then simplifies to the steady state ($x_{\infty}$):
:::{math}
:enumerated: false
x_{\infty} = \lim_{t\to\infty} \left( a^{t}x_{0} + \frac{1-a^{t}}{1-a}c \right) = 0 \cdot x_{0} + \frac{1-0}{1-a}c = \frac{c}{1-a}
:::

> This confirms that when $|a| < 1$, the series is **stable** and converges to $x_{\infty}$ regardless of the starting value $x_0$.

* **If $|a| > 1$**: As $t \to \infty$, the term $a^{t}$ diverges toward $\pm\infty$. 
    In this case, the series is **unstable** and moves further away from the steady state over time (unless $x_0$ is exactly equal to the steady state value).

## General Solution to First-Order Difference Equations

The general solution can be found by combining the **complementary solution** and the **particular solution**. 

:::{math}
:enumerated: false
\text{General solution} = \text{Complementary solution} + \text{Particular solution}
:::


And as we will see later, this method is especially useful for solving higher-order equations later.

#### 1. The Complementary Solution ($x_c$)

The complementary solution is obtained by considering the **homogeneous equation**, which means setting the constant term $c$ to zero:

:::{math}
:enumerated: false
x_{t} - ax_{t-1} = 0
:::

We assume a solution of the form $x_{t} = Ak^{t}$. Substituting this into the homogeneous equation gives:

:::{math}
:enumerated: false
Ak^{t} - aAk^{t-1} = 0
:::

Dividing by $Ak^{t-1}$ gives $k = a$. Therefore, the complementary solution is:

:::{math}
:enumerated: false
x_{t} = Aa^{t}
:::

* **$A$** is a constant to be determined by initial conditions.
* This part of the function reflects the **deviation** from the steady state.

#### 2. The Particular Solution ($x_p$)

The particular solution is any solution that satisfies the non-homogeneous difference equation. It is most convenient to use the **steady state value** ($x_{\infty}$) as the particular solution. 

As derived previously, for $a \neq 1$:

:::{math}
:enumerated: false
x_{p} = \frac{c}{1 - a}
:::

* The particular solution represents the **inter-temporal equilibrium** or steady state.

::: {.callout-note}
**Intuition**

The steady state is the value of $x$ that reproduces itself over time. If the system reaches $x^*$, then $x_{t}=x_{t-1}=x^*$ and there is no further change.
:::

#### 3. The General Solution

The general solution is the sum of the complementary and particular solutions:

:::{math}
:enumerated: false
x_{t} = Aa^{t} + \frac{c}{1 - a}
:::

#### 4. The Definite Solution

To find the specific value for $A$, we use the initial condition at $t = 0$:

:::{math}
:enumerated: false
x_{0} = Aa^{0} + \frac{c}{1 - a}
:::

:::{math}
:enumerated: false
A = x_{0} - \frac{c}{1 - a}
:::

Substituting this value of $A$ back into the general solution gives the **definite solution**:

:::{math}
:enumerated: false
x_{t} = \left( x_{0} - \frac{c}{1 - a} \right) a^{t} + \frac{c}{1 - a}
:::

Which can be rearranged to match the result from the iteration method:

:::{math}
:enumerated: false
x_{t} = a^{t}x_{0} + \frac{1 - a^{t}}{1 - a}c
:::

> Note that this is exactly the same Equation {eq}`defn-solution-1` derived previously.
---

### Stability Condition

The system is considered **stable** as long as the complementary part $Aa^{t} \to 0$ as $t \to \infty$

* If **$|a| < 1$**, the time path will converge to the equilibrium (steady state).
* If **$|a| > 1$**, the time path will diverge away from the equilibrium.
* Also, if **$a = 1$**, we have $x_t = x_0 + ct$.

::: {.callout-important}
**Common Pitfall**

Do not confuse **existence of a steady state** with **stability**.

A steady state may exist even when $|a|>1$, but the system will move away from it unless the initial value is exactly equal to $x^*$.
:::

### Numerical Example

This example follows the general solution method by combining the complementary and particular solutions to find a definite solution for a specific system.

**Given Equation:**

:::{math}
:enumerated: false
x_{t} = -7x_{t-1} + 16
:::

**Initial Condition:**
:::{math}
:enumerated: false
x_{0} = 5
:::

---

### Step 1: Identify Parameters

From the general form $x_t = ax_{t-1} + c$, we identify:

* $a = -7$
* $c = 16$
* $x_0 = 5$

### Step 2: Find the Particular Solution ($x_p$)

The particular solution represents the steady state ($x_{\infty}$). For $a \neq 1$, it is calculated as:

:::{math}
:enumerated: false
x_{p} = \frac{c}{1 - a} = \frac{16}{1 - (-7)} = \frac{16}{8} = 2
:::

### Step 3: Find the Complementary Solution ($x_c$)

The complementary solution takes the form $Aa^t$. Using our value for $a$:

:::{math}
:enumerated: false
x_{c} = A(-7)^{t}
:::

### Step 4: Form the General Solution

The general solution is the sum of the complementary and particular solutions:

:::{math}
:enumerated: false
x_{t} = A(-7)^{t} + 2
:::

### Step 5: Solve for the Constant $A$ (Definite Solution)

Use the initial condition $x_0 = 5$ at $t = 0$:

:::{math}
:enumerated: false
5 = A(-7)^{0} + 2
:::

:::{math}
:enumerated: false
5 = A(1) + 2
:::

:::{math}
:enumerated: false
A = 5 - 2 = 3
:::

Substituting $A$ back into the general solution gives the **Definite Solution**:

:::{math}
:enumerated: false
x_{t} = 3(-7)^{t} + 2
:::

---

### Verification

We can verify the solution by calculating the first two terms and checking them against the original difference equation:

* **At $t=0$:** $x_0 = 3(-7)^0 + 2 = 5$ (Matches initial condition).
* **At $t=1$:** $x_1 = 3(-7)^1 + 2 = -21 + 2 = -19$.

Now, substitute $x_0 = 5$ into the original difference equation to see if it produces $x_1 = -19$:

:::{math}
:enumerated: false
x_1 = -7(5) + 16 = -35 + 16 = -19
:::

The values match, confirming the solution is correct.

### Stability Analysis

Because $|a| = |-7| = 7$, which is **greater than 1**, the time path is **unstable** and will diverge. Additionally, because $a < 0$, the path will be **oscillatory**, jumping between positive and negative deviations from the steady state.

### Stability Conditions

We know have learnt that the difference equation of the form

:::{math}
:enumerated: false
x_t = a x_{t-1} + c
:::

has as its solution

:::{math}
:enumerated: false
x_t = \left( x_0 - \frac{1}{1 - a} \right) a^t + \frac{1}{1 - a} c
:::

which can be seen as essentially made up of two parts:

That is

:::{math}
:enumerated: false
\left( x_0 - \frac{1}{1 - a} \right) = A \quad \text{and} \quad \frac{1}{1 - a} c = p,
::::

and hence we have a general form $x_t = A a^t + p$.

Needless to say,

:::{math}
:enumerated: false
A = \left( x_0 - \frac{1}{1 - a} \right)
:::

forms the complementary function reflecting the deviation from equilibrium or the steady state, while

:::{math}
:enumerated: false
p = \frac{1}{1 - a} c
:::

the particular function represents the inter-temporal equilibrium level of $x_t$.

Hence, the system is stable as long as $A a^t \to 0$ as $t \to \infty$ (so the dynamics of the system depends on $a$.

Assuming $A = 1$ and $p = 0$ for now, $a^t$ will generate seven different paths depending on $a$.

```python
import numpy as np
import matplotlib.pyplot as plt

a_values = [1.5, 1, 0.5, 0, -0.5, -1, -1.5]
titles = [
    "(a) $a > 1$: Monotonic Divergence",
    "(b) $a = 1$: Constant Path",
    "(c) $0 < a < 1$: Monotonic Convergence",
    "(d) $a = 0$: Immediate Equilibrium",
    "(e) $-1 < a < 0$: Oscillatory Convergence",
    "(f) $a = -1$: Constant Oscillation",
    "(g) $a < -1$: Oscillatory Divergence"
]

n = np.arange(0, 10)

fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.flatten()

for i, a in enumerate(a_values):
    y = a**n
    
    # Removed 'o' from the format string to get rid of dots
    axes[i].plot(n, y, '-', drawstyle='steps-post', linewidth=2)
    
    axes[i].set_title(titles[i], fontsize=11)
    axes[i].set_xlabel("Time ($t$)")
    axes[i].set_ylabel("$a^t$")
    axes[i].axhline(0, color='black', lw=0.8, alpha=0.4)
    axes[i].grid(True, alpha=0.2)

axes[-1].axis('off')
plt.suptitle("Time Paths", fontsize=14)
plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

On the y-axis we have $a^t$ and on the x-axis we have $t$.

**In sum**
- If $|a| > 1$, the time path will diverge  
- If $|a| < 1$, the time path will converge to equilibrium  

**Also**
- If $a < 0$, the time path will oscillate  
- If $a > 0$, the time path will not oscillate  

**Scaling and Shifts**
- If $A \neq 1$, the diagram shown above will simply scale up/down depending on the magnitude of $a^t$. The pattern will remain essentially the same.  
- If $A = -1$, we will get a mirror image with respect to the horizontal axis.  
- If $c \neq 0$, the vertical intercept is affected (graph shifts up/down).  

## Phase Diagrams

We can draw phase diagrams for difference equations (instead of path diagrams) to illustrate their characteristics. Phase diagrams are used to illustrate the dynamic path of a variable generated by a difference equation and are especially useful for getting some qualitative sense of the solution of a nonlinear difference equations.  Below are phase diagrams for the same equations which we studied above.

```python
# import numpy as np
# import matplotlib.pyplot as plt

# 1. Parameters
a, c, x0 = 0.5, 300, 400
f = lambda x: a * x + c
x_star = c / (1 - a)

# 2. Setup Plot Data
x_vals = np.linspace(350, 650, 200)

# 3. Generate Cobweb Path Logic
path_x, path_y = [x0], [x0] 
for _ in range(6):
    # Vertical move
    path_x.append(path_x[-1])
    path_y.append(f(path_x[-1]))
    # Horizontal move
    path_x.append(path_y[-1])
    path_y.append(path_y[-1])

# 4. Plotting
plt.figure(figsize=(6,6))

# Main Lines
plt.plot(x_vals, x_vals, 'k--', alpha=0.6, label=r"$x_t = x_{t-1}$")
plt.plot(x_vals, f(x_vals), 'b', lw=2, label=r"$x_t = 0.5x_{t-1} + 300$")

# Cobweb Path with arrows
for i in range(1, len(path_x)):
    plt.annotate("",
                 xy=(path_x[i], path_y[i]),
                 xytext=(path_x[i-1], path_y[i-1]),
                 arrowprops=dict(arrowstyle="->", color="green"))

# Equilibrium Point
plt.scatter(x_star, x_star, color='red', s=80, zorder=5,
            label=f"Equilibrium ($x^*={int(x_star)}$)")

# Formatting
plt.title(f"Phase Diagram: Convergence ($a={a}$)", fontsize=13)
plt.xlabel("$x_{t-1}$", fontsize=11)
plt.ylabel("$x_t$", fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()

# Tight axis limits to remove whitespace
plt.xlim(350, 650)
plt.ylim(300, 650)

plt.tight_layout()
plt.show()
```

```python
# import numpy as np
# import matplotlib.pyplot as plt

# Parameters
a, c, x0 = -0.5, 300, 400
f = lambda x: a * x + c
x_star = c / (1 - a)   # equilibrium point

# Range for plotting
x_vals = np.linspace(0, 450, 200)

# Cobweb path
steps = 6
path_x, path_y = [x0], [x0]
for _ in range(steps):
    # Vertical move
    path_x.append(path_x[-1])
    path_y.append(f(path_x[-1]))
    # Horizontal move
    path_x.append(path_y[-1])
    path_y.append(path_y[-1])

# Plot
plt.figure(figsize=(6,6))
plt.plot(x_vals, x_vals, 'k--', label=r"$x_t = x_{t-1}$")
plt.plot(x_vals, f(x_vals), 'b-', lw=2, label=r"$x_t = -0.5x_{t-1} + 300$")

# Cobweb path with arrows
for i in range(1, len(path_x)):
    plt.annotate("",
                 xy=(path_x[i], path_y[i]),
                 xytext=(path_x[i-1], path_y[i-1]),
                 arrowprops=dict(arrowstyle="->", color="green"))

# Equilibrium point
plt.scatter(x_star, x_star, color='red', s=80, zorder=5,
            label=f"Equilibrium ($x^*={int(x_star)}$)")

# Reference lines (only equilibrium)
plt.axhline(x_star, color='gray', linestyle='--', linewidth=0.7)
plt.axvline(x_star, color='gray', linestyle='--', linewidth=0.7)

# Labels and formatting
plt.title("Phase Diagram: Oscillatory Convergence to $x^* = 200$", fontsize=13)
plt.xlabel(r"$x_{t-1}$")
plt.ylabel(r"$x_t$")
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)

plt.xlim(0, 450)
plt.ylim(0, 450)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
a, c, x0 = -2, 900, 400
f = lambda x: a * x + c
x_star = c / (1 - a)   # equilibrium point

# Range for plotting
x_vals = np.linspace(0, 800, 400)

# Cobweb path
steps = 6
path_x, path_y = [x0], [x0]
for _ in range(steps):
    # Vertical move
    path_x.append(path_x[-1])
    y_val = f(path_x[-1])
    # Clip at 0 so it stops at (700,0) instead of -500
    path_y.append(max(y_val, 0))
    # Horizontal move
    path_x.append(path_y[-1])
    path_y.append(path_y[-1])

# Plot
plt.figure(figsize=(7,7))
plt.plot(x_vals, x_vals, 'k--', label=r"$x_t = x_{t-1}$", zorder=2)
plt.plot(x_vals, f(x_vals), 'b-', lw=2, label=r"$x_t = -2x_{t-1} + 900$", zorder=2)

# Cobweb path with arrows (behind legend)
for i in range(1, len(path_x)):
    plt.annotate("",
                 xy=(path_x[i], path_y[i]),
                 xytext=(path_x[i-1], path_y[i-1]),
                 arrowprops=dict(arrowstyle="->", color="green"),
                 zorder=1)  # arrows behind

# Equilibrium point
plt.scatter(x_star, x_star, color='red', s=80, zorder=3,
            label=f"Equilibrium ($x^*={int(x_star)}$)")

# Reference lines
plt.axhline(x_star, color='gray', linestyle='--', linewidth=0.7, zorder=1)
plt.axvline(x_star, color='gray', linestyle='--', linewidth=0.7, zorder=1)

# Labels and formatting
plt.title("Phase Diagram: Oscillatory Divergence from $x^* = 300$", fontsize=13)
plt.xlabel(r"$x_{t-1}$")
plt.ylabel(r"$x_t$")
plt.legend()  # no zorder here
plt.grid(True, linestyle=':', alpha=0.6)

# Expanded limits
plt.xlim(0, 800)
plt.ylim(0, 800)

plt.tight_layout()
plt.show()
```

## Second-Order Difference Equations

We can write a second–order difference equation as:

:::{math}
:enumerated: false
x_t = a x_{t-1} + b x_{t-2} + c
:::

To find the general solution, we need both the **particular solution** (steady state) and the **complementary solution** (homogeneous part).

### Particular Solution (Steady State)

We look for the long-run value $x_\infty$:

:::{math}
:enumerated: false
\lim_{t \to \infty} x_t = \lim_{t \to \infty} (a x_{t-1} + b x_{t-2} + c)
:::

Assuming convergence:

:::{math}
:enumerated: false
x_\infty = a x_\infty + b x_\infty + c
:::

:::{math}
:enumerated: false
x_\infty (1 - a - b) = c
:::

Thus:

:::{math}
:enumerated: false
x_\infty = \frac{c}{1 - a - b}
:::

⚠️ If $a + b = 1$, then no finite steady state exists (unless $x_{t-1} = x_{t-2} = 0$).

### Complementary Solution (Homogeneous Case)

Consider the homogeneous equation:

:::{math}
:enumerated: false
x_t - a x_{t-1} - b x_{t-2} = 0
:::

Suppose a solution of the form:

:::{math}
:enumerated: false
x_t = A k^t
:::

Then:

:::{math}
:enumerated: false
x_{t-1} = A k^{t-1}, \quad x_{t-2} = A k^{t-2}
:::

Substitute:

:::{math}
:enumerated: false
A k^t - a A k^{t-1} - b A k^{t-2} = 0
:::

Divide through by $A$:

:::{math}
:enumerated: false
k^t - a k^{t-1} - b k^{t-2} = 0
:::

Factor:

:::{math}
:enumerated: false
(k^2 - a k - b) k^{t-2} = 0
:::

So the **characteristic equation** is:

:::{math}
:enumerated: false
k^2 - a k - b = 0
:::

### Roots of the Characteristic Equation

:::{math}
:enumerated: false
k_{1,2} = \frac{a \pm \sqrt{a^2 + 4b}}{2}
:::

or equivalently:

:::{math}
:enumerated: false
k_{1,2} = \frac{a}{2} \pm \sqrt{\frac{1}{4}a^2 + b}
:::

---

### ✅ Summary

- **Particular solution (steady state):** $x_\infty = \displaystyle\frac{c}{1 - a - b}$  
- **Complementary solution:** determined by the roots of $k^2 - a k - b = 0$  
- **The general solution is the sum of both.**

---

### Cases for Second-Order Difference Equations

There are three possible sets of values for $k_1$ and $k_2$, depending on the parameters $a$ and $b$.


**Case 1: Real and Distinct Roots**

Condition:

:::{math}
:enumerated: false
a^2 + 4b > 0, \quad k_1 \neq k_2
:::

Solution:  

:::{math}
:enumerated: false
x_t = A k_1^t + B k_2^t
:::

where $A$ and $B$ are constants determined by initial conditions.

**Case 2: Repeated Root**

Condition:  

:::{math}
:enumerated: false
a^2 + 4b = 0, \quad k_1 = k_2
:::

Solution:  

:::{math}
:enumerated: false
x_t = (A + Bt) k_1^t
:::

where $A, B$ are constants.

**Case 3: Complex Roots**

Condition:  

:::{math}
:enumerated: false
a^2 + 4b < 0
:::

Roots:  

:::{math}
:enumerated: false
k_{1,2} = r(\cos \theta \pm i \sin \theta), \quad r > 0, \quad \theta \in (-\pi, \pi)
:::

Solution:  

:::{math}
:enumerated: false
x_t = A r^t \cos(t\theta + \omega)
:::

where $A, \omega$ are constants determined by initial conditions.

---

### ✅ Summary

:::{math}
:enumerated: false
x_t = \big(\text{complementary solution depending on root case}\big) + x_{\infty}
:::

- **Case 1 (real distinct roots):**  
:::{math}
:enumerated: false
  x_t = A k_1^t + B k_2^t + x_{\infty}
:::

- **Case 2 (repeated root):**  
:::{math}
:enumerated: false
  x_t = (A + Bt) k_1^t + x_{\infty}
:::

- **Case 3 (complex roots):**  
:::{math}
:enumerated: false
  x_t = A r^t \cos(t\theta + \omega) + x_{\infty}
:::

---

### Stability Conditions

- For **real roots**, convergence requires the **dominant root** (largest \(|k|\)) to satisfy:  
:::{math}
:enumerated: false
  |k| < 1
:::

- For **complex roots**, convergence requires the modulus:  

:::{math}
:enumerated: false
  r < 1
:::

- Otherwise, the system **diverges**.


### Worked Example: Solving a Second-Order Difference Equation

Consider:

:::{math}
:enumerated: false
x_t - 10x_{t-1} + 16x_{t-2} = 14
:::

**Step 1: Particular Solution**

:::{math}
:enumerated: false
x_p = \frac{14}{1 - 10 - (-16)} = \frac{14}{7} = 2
:::

**Step 2: Characteristic Equation**

:::{math}
:enumerated: false
k_1, k_2 = \frac{10 \pm \sqrt{100 + 4(-16)}}{2}
:::

:::{math}
:enumerated: false
= \frac{10 \pm \sqrt{36}}{2}
:::

:::{math}
:enumerated: false
= \frac{10 \pm 6}{2} \quad \Rightarrow \quad k_1 = 2, \; k_2 = 8
:::

**Step 3: General Solution**

Since the roots are real and distinct (Case 1):

:::{math}
:enumerated: false
x_t = A(2)^t + B(8)^t + 2
:::

**Step 4: Apply Initial Conditions**

Suppose $x_0 = 10$ and $x_1 = 36$.

- At $t=0$:

:::{math}
:enumerated: false
x_0 = A(2^0) + B(8^0) + 2 = A + B + 2
:::

:::{math}
:enumerated: false
A + B + 2 = 10 \quad \Rightarrow \quad A + B = 8
:::

- At $t=1$:

:::{math}
:enumerated: false
x_1 = A(2) + B(8) + 2 = 2A + 8B + 2
:::

:::{math}
:enumerated: false
2A + 8B + 2 = 36 \quad \Rightarrow \quad 2A + 8B = 34
:::

**Step 5: Solving Simultaneously**

From the system:

:::{math}
:enumerated: false
A + B = 8
:::

:::{math}
:enumerated: false
2A + 8B = 34
:::

Solving gives:

:::{math}
:enumerated: false
A = 5, \quad B = 3
:::

**Final Definite Solution**

:::{math}
:enumerated: false
x_t = 5(2)^t + 3(8)^t + 2
:::

### Example 2: Complex Roots in a Second-Order Difference Equation

Consider:

:::{math}
:enumerated: false
x_t = 0.6x_{t-1} + 0.8x_{t-2}
:::

**Step 1: Particular Solution**

Since there is no constant term ($c=0$$), the steady state is:

:::{math}
:enumerated: false
x_\infty = \frac{c}{1 - a - b} = 0
:::

**Step 2: Characteristic Equation**

:::{math}
:enumerated: false
k^2 - 0.6k - 0.8 = 0
:::

:::{math}
:enumerated: false
k = \frac{0.6 \pm \sqrt{0.36 + 3.2}}{2}
= \frac{0.6 \pm \sqrt{3.56}}{2}
:::

:::{math}
:enumerated: false
= \frac{0.6 \pm 1.887}{2}
:::

:::{math}
:enumerated: false
k_1 \approx 1.243, \quad k_2 \approx -0.643
:::

**Step 3: Complex Root Example**

To illustrate oscillatory behavior, let’s adjust parameters so that:

:::{math}
:enumerated: false
a = 0.4, \quad b = 0.8
:::

Characteristic equation:

:::{math}
:enumerated: false
k^2 - 0.4k - 0.8 = 0
:::

Discriminant:

:::{math}
:enumerated: false
(0.4)^2 + 4(0.8) = 0.16 + 3.2 = 3.36 < 0
:::

So roots are complex.

**Step 4: Express Roots in Polar Form**

:::{math}
:enumerated: false
k_{1,2} = r(\cos \theta \pm i \sin \theta)
:::

where:

- $r = \sqrt{b} = \sqrt{0.8} \approx 0.894$  
- $\theta = \arccos\left(\frac{a}{2r}\right)$

**Step 5: General Solution**

:::{math}
:enumerated: false
x_t = A r^t \cos(t\theta + \omega)
:::

This represents **oscillatory convergence** because $r < 1$.

**Step 6: Apply Initial Conditions (Example)**

Suppose:

:::{math}
:enumerated: false
x_0 = 1, \quad x_1 = 0
:::

We can solve for constants $A, \omega$ using these starting values. The resulting solution will oscillate around zero, with amplitude shrinking at rate $r^t$.

### ✅ Final Form

For complex roots:

:::{math}
:enumerated: false
x_t = A r^t \cos(t\theta + \omega) + x_\infty
:::

- Oscillatory behavior comes from the cosine term.  
- Convergence/divergence depends on $r$.  
- If $r < 1$, oscillations shrink toward equilibrium.  
- If $r > 1$, oscillations grow and diverge.


**More concretely**

Suppose:

:::{math}
:enumerated: false
x_0 = 1, \quad x_1 = 0
:::

- At $t=0$:

:::{math}
:enumerated: false
x_0 = A \cos(\omega) = 1
:::

- At $t=1$:

:::{math}
:enumerated: false
x_1 = A r \cos(\theta + \omega) = 0
:::

This implies:

:::{math}
:enumerated: false
\cos(\theta + \omega) = 0 \quad \Rightarrow \quad \theta + \omega = \frac{\pi}{2}
:::

So:

:::{math}
:enumerated: false
\omega = \frac{\pi}{2} - \theta \approx \frac{\pi}{2} - 1.345 \approx 0.226 \,\text{rad}
:::

And from $x_0 = 1$:

:::{math}
:enumerated: false
A \cos(\omega) = 1 \quad \Rightarrow \quad A = \frac{1}{\cos(0.226)} \approx 1.026
:::

**Final Definite Solution**

:::{math}
:enumerated: false
x_t \approx 1.026 \cdot (0.894)^t \cos(1.345t + 0.226)
:::

---

### ✅ Interpretation
- The sequence oscillates because of the cosine term.  
- The amplitude shrinks at rate $0.894^t$.  
- The equilibrium is $x_\infty = 0$.  

---

We can plot this:

```python
# import numpy as np
# import matplotlib.pyplot as plt

# Parameters
A = 1.026
r = 0.894
theta = 1.345
omega = 0.226

# Time steps
t = np.arange(0, 21)

# Compute x_t values
x_t = A * (r ** t) * np.cos(theta * t + omega)

# Plot
plt.style.use('seaborn-v0_8')
plt.figure(figsize=(10, 6))
plt.plot(t, x_t, marker='o', linestyle='-', color='royalblue')
plt.title(r'Oscillatory Solution of $x_t = 0.4x_{t-1} + 0.8x_{t-2}$', fontsize=14)
plt.xlabel('Time $t$', fontsize=12)
plt.ylabel(r'$x_t$', fontsize=12)
plt.grid(True)
plt.tight_layout()
plt.show()
```

::: {.callout-note}
**Key Idea**

With more than one variable, difference equations must be solved **jointly**. The evolution of each variable depends on the others, so we move from scalar dynamics to **system dynamics**.
:::
